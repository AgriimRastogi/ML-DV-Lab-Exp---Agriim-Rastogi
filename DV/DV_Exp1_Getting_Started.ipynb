# Experiment 1: Getting Started with Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]
print(df.head(10))
print(df.describe())
```

```python
avg_petal = df.groupby('species')['petal length (cm)'].mean()
print(avg_petal)
plt.bar(avg_petal.index, avg_petal.values, color=['skyblue', 'lightgreen', 'salmon'])
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
plt.plot(df['sepal length (cm)'][:30], marker='o', color='blue')
plt.title('Sepal Length for First 30 Samples')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```
