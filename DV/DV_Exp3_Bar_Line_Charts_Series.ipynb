# Experiment 3: Creating Bar and Line Charts for Series Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Sales': [12000, 14500, 13800, 16200, 18000, 17500, 19400, 21000, 20200, 23500, 26000, 29500]
}
cat_data = {
    'Category': ['Laptops', 'Smartphones', 'Tablets', 'Accessories', 'Monitors'],
    'Revenue': [85000, 72000, 34000, 28000, 46000]
}
df_monthly = pd.DataFrame(sales_data)
df_cat = pd.DataFrame(cat_data)
print(df_monthly.head())
print(df_cat)
```

```python
bars = plt.bar(df_cat['Category'], df_cat['Revenue'], color='teal')
for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, h + 1200, str(h), ha='center')
plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue ($)')
plt.ylim(0, 98000)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df_monthly['Month'], df_monthly['Sales'], marker='o', color='darkblue', linewidth=2)
for i in range(len(df_monthly)):
    plt.text(i, df_monthly['Sales'][i] + 500, str(df_monthly['Sales'][i]), ha='center', fontsize=8)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True)
plt.show()
```
