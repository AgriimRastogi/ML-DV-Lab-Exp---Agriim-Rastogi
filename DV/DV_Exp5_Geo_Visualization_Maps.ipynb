# Experiment 5: Geo-Visualization with Maps

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
geo_data = {
    'City': ['New Delhi', 'Mumbai', 'Bengaluru', 'New York', 'London', 'Tokyo', 'Sydney', 'Toronto', 'Paris', 'Singapore'],
    'Country': ['India', 'India', 'India', 'USA', 'UK', 'Japan', 'Australia', 'Canada', 'France', 'Singapore'],
    'Latitude': [28.61, 19.07, 12.97, 40.71, 51.50, 35.67, -33.86, 43.65, 48.85, 1.35],
    'Longitude': [77.20, 72.87, 77.59, -74.00, -0.12, 139.65, 151.20, -79.38, 2.35, 103.81],
    'Sales': [42000, 58000, 51000, 84000, 63000, 71000, 39000, 47000, 52000, 44000]
}
df = pd.DataFrame(geo_data)
print(df)
```

```python
plt.figure(figsize=(12, 6))
sc = plt.scatter(df['Longitude'], df['Latitude'], s=df['Sales'] / 120, c=df['Sales'], cmap='YlOrRd', alpha=0.75, edgecolors='black')
for i in range(len(df)):
    plt.text(df['Longitude'][i] + 2, df['Latitude'][i] + 1.5, df['City'][i], fontsize=8)
plt.colorbar(sc, label='Sales ($)')
plt.title('Global City-Wise Sales Symbol Map')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.xlim(-110, 175)
plt.ylim(-50, 65)
plt.grid(True)
plt.show()
```

```python
country_sales = df.groupby('Country')['Sales'].sum().sort_values()
plt.barh(country_sales.index, country_sales.values, color='teal')
plt.title('Total Sales by Country')
plt.xlabel('Sales ($)')
plt.ylabel('Country')
plt.show()
```
