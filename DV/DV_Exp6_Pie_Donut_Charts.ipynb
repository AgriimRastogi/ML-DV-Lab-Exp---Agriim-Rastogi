# Experiment 6: Pie and Donut Charts for Category Proportion

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
olympics = {
    'Country': ['USA', 'China', 'Japan', 'Great Britain', 'France', 'Australia', 'India'],
    'Gold': [40, 40, 20, 14, 16, 18, 1],
    'Silver': [44, 27, 12, 22, 26, 19, 2],
    'Bronze': [42, 24, 13, 29, 22, 16, 4]
}
df = pd.DataFrame(olympics)
df['Total'] = df['Gold'] + df['Silver'] + df['Bronze']
print(df)
```

```python
plt.figure(figsize=(7, 7))
plt.pie(df['Total'], labels=df['Country'], autopct='%1.1f%%', startangle=140)
plt.title('Total Olympic Medal Share by Country')
plt.show()
```

```python
plt.figure(figsize=(7, 7))
plt.pie(df['Gold'], labels=df['Country'], autopct='%1.1f%%', startangle=140, wedgeprops={'width': 0.4})
plt.title('Gold Medal Share by Country (Donut Chart)')
plt.show()
```
